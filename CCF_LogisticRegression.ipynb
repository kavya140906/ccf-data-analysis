```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# Read the CSV with full path (update this path if needed)
try:
    credit_card_data = pd.read_csv("C:/Users/kavya sri CM/Downloads/creditcard.csv")
    print("Data loaded successfully!")
except FileNotFoundError:
    print("CreditCard.csv not found.please check the file path.")
```

```python
print("\nDataset Summary:")
print(credit_card_data.info())
```

```python
if credit_card_data.isnull().sum().sum()>0:
    print("Missing values found! Filling with forward fill.")
    credit_card_data.fillna(method = "ffill",inplace = True)
```

```python
import warnings

warnings.filterwarnings("ignore")
# Visualize class imbalance
plt.figure(figsize=(6,4))
sns.countplot(x="Class",data=credit_card_data,palette = "Set2")
plt.title("Fraud vs Non-Fraud Transactions")
plt.xlabel("Transacton class(0 = Non-Fraud, 1=Fraud)")
plt.ylabel("Count")
plt.grid(axis = "y", linestyle ="--", alpha = 0.7)
plt.tight_layout()
plt.show()
```

```python
import warnings

warnings.filterwarnings("ignore")
# Correlation heatmap (Zoomed into top correlation)
plt.figure(figsize=(20,12))
corr = credit_card_data.corr()
```

```python
# Sort  correlattions with respect to 'class'
sorted_corr = corr["Class"].abs().sort_values(ascending=False)
sns.heatmap(corr[sorted_corr.index][:20], cmap="coolwarm",annot=False,linewidth=0.5)
plt.title("Top correlation heatmap with class")
plt.tight_layout()
plt.show()
```
